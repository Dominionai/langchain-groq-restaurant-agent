```python
llm = ChatGroq(temperature=0.6, model_name="llama-3.3-70b-versatile")

prompt_template_name = PromptTemplate(
    input_variables=['cuisine'],
    template="I want to open a restaurant for {cuisine} food. Suggest a fancy name for this. Return only the name, nothing else."
)

prompt_template_items = PromptTemplate(
    input_variables=['restaurant_name'],
    template="Suggest some menu items for {restaurant_name}. Return as a comma separated string."
)

# LCEL pipes
name_chain = prompt_template_name | llm | parser
food_items_chain = prompt_template_items | llm | parser
```

#### Simple Sequential Chain

```python
# SimpleSequentialChain equivalent in LCEL: chain the two pipes together
from langchain_core.runnables import RunnableLambda

simple_chain = (
    prompt_template_name
    | llm
    | parser
    | RunnableLambda(lambda name: {"restaurant_name": name})
    | prompt_template_items
    | llm
    | parser
)

content = simple_chain.invoke({"cuisine": "Nigeria"})
print(content)
```

#### Sequential Chain

```python
llm = ChatGroq(temperature=0.7, model_name="llama-3.3-70b-versatile")

prompt_template_name = PromptTemplate(
    input_variables=['cuisine'],
    template="I want to open a restaurant for {cuisine} food. Suggest a fancy name for this. Return only the name, nothing else."
)

name_chain = prompt_template_name | llm | parser
```

```python
llm = ChatGroq(temperature=0.7, model_name="llama-3.3-70b-versatile")

prompt_template_items = PromptTemplate(
    input_variables=['restaurant_name'],
    template="Suggest some menu items for {restaurant_name}. Return as a comma separated string."
)

food_items_chain = prompt_template_items | llm | parser
```

```python
from langchain_core.runnables import RunnablePassthrough

# SequentialChain equivalent: pass both restaurant_name and menu_items as dict
sequential_chain = (
    RunnablePassthrough.assign(
        restaurant_name=name_chain
    )
    | RunnablePassthrough.assign(
        menu_items=lambda x: food_items_chain.invoke({"restaurant_name": x["restaurant_name"]})
    )
)
```

```python
sequential_chain.invoke({"cuisine": "Nigeria"})
```

#### serpapi and llm-math tool

```python
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage, ToolMessage
from langchain_community.tools import GoogleSerperRun
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

# Use llama-3.1-8b-instant — most stable tool-calling model on Groq
agent_llm = ChatGroq(temperature=0, model_name='llama-3.1-8b-instant')

# Google Search tool (replaces serpapi)
search_tool = GoogleSerperRun(api_wrapper=GoogleSerperAPIWrapper())

# llm-math equivalent
@tool
def calculator(expression: str) -> str:
    """Evaluate a math expression e.g. '25878900000000 + 5'. Must be a plain Python expression."""
    try:
        return str(eval(expression, {'__builtins__': {}}, {}))
    except Exception as e:
        return f'Error: {e}'

# agent runner — no langchain.agents needed
def run_agent(input_text, tools):
    tools_map = {t.name: t for t in tools}
    llm_with_tools = agent_llm.bind_tools(tools)
    messages = [HumanMessage(content=input_text)]
    while True:
        response = llm_with_tools.invoke(messages)
        messages.append(response)
        if not response.tool_calls:
            break
        for tc in response.tool_calls:
            tool_result = tools_map[tc['name']].invoke(tc['args'])
            messages.append(ToolMessage(content=str(tool_result), tool_call_id=tc['id']))
    return response.content

# Run with Google Search + calculator
tools = [search_tool, calculator]
print(run_agent('What was the GDP of US in 2026? Then add 5 to that number.', tools))
```

#### Wikipedia and llm-math tool

```python
wiki_tool = WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper())

tools = [wiki_tool, calculator]
print(tools[0].run('When was Elon Musk born? What is his current age?', tools))
```

## Memory

```python
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

llm = ChatGroq(temperature=0.7, model_name='llama-3.3-70b-versatile')

memory_prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful restaurant name generator.'),
    MessagesPlaceholder(variable_name='history'),
    ('human', '{input}'),
])

memory_chain = memory_prompt | llm | parser

store = {}

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]

chain_with_history = RunnableWithMessageHistory(
    memory_chain,
    get_session_history,
    input_messages_key='input',
    history_messages_key='history',
)

name = chain_with_history.invoke(
    {'input': 'I want to open a Mexican restaurant. Suggest a fancy name.'},
    config={'configurable': {'session_id': 'resto-session'}}
)
print(name)
```

```python
name = chain_with_history.invoke(
    {'input': 'I want to open a Nigeria restaurant. Suggest a fancy name.'},
    config={'configurable': {'session_id': 'resto-session'}}
)
print(name)
```

#### ConversationChain

```python
convo_prompt = ChatPromptTemplate.from_messages([
    ('system', 'The following is a friendly conversation between a human and an AI. '
               'The AI is talkative and provides lots of specific details from its context. '
               'If the AI does not know the answer, it truthfully says it does not know.'),
    MessagesPlaceholder(variable_name='history'),
    ('human', '{input}'),
])

convo_chain = convo_prompt | ChatGroq(temperature=0.7, model_name='llama-3.3-70b-versatile') | parser

convo_store = {}

def get_convo_history(session_id: str):
    if session_id not in convo_store:
        convo_store[session_id] = InMemoryChatMessageHistory()
    return convo_store[session_id]

convo = RunnableWithMessageHistory(
    convo_chain,
    get_convo_history,
    input_messages_key='input',
    history_messages_key='history',
)

# Show the prompt structure (equivalent to print(convo.prompt.template))
print(convo_prompt.messages)
```

```python
convo.invoke(
    {'input': 'Who won the first cricket world cup?'},
    config={'configurable': {'session_id': 'convo-1'}}
)
```

```python
convo.invoke(
    {'input': 'How much is 5+5?'},
    config={'configurable': {'session_id': 'convo-1'}}
)
```

```python
convo.invoke(
    {'input': 'Who was the captain of the winning team?'},
    config={'configurable': {'session_id': 'convo-1'}}
)
```

```python
# Equivalent to print(convo.memory.buffer)
for msg in convo_store['convo-1'].messages:
    print(f'{msg.type}: {msg.content}')
```

#### ConversationBufferWindowMemory

```python
from langchain_core.runnables import RunnableLambda

K = 1  # keep only the last K human+AI turn pairs

def trim_history(input_dict):
    history = input_dict.get('history', [])
    input_dict['history'] = history[-(K * 2):]
    return input_dict

window_chain = (
    RunnableLambda(trim_history)
    | convo_prompt
    | ChatGroq(temperature=0.7, model_name='llama-3.3-70b-versatile')
    | parser
)

window_store = {}

def get_window_history(session_id: str):
    if session_id not in window_store:
        window_store[session_id] = InMemoryChatMessageHistory()
    return window_store[session_id]

convo_window = RunnableWithMessageHistory(
    window_chain,
    get_window_history,
    input_messages_key='input',
    history_messages_key='history',
)

convo_window.invoke(
    {'input': 'Who won the first cricket world cup?'},
    config={'configurable': {'session_id': 'window-1'}}
)
```

```python
convo_window.invoke(
    {'input': 'How much is 5+5?'},
    config={'configurable': {'session_id': 'window-1'}}
)
```

```python
convo_window.invoke(
    {'input': 'Who was the captain of the winning team?'},
    config={'configurable': {'session_id': 'window-1'}}
)
```
